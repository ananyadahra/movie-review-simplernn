```python
import numpy as np
import tensorflow as tf
from keras.datasets import imdb
from keras.preprocessing import sequence
from keras.models import load_model
```

```python
word_index = imdb.get_word_index()
reversed_word_index = {value: key for (key, value) in word_index.items()}
```

```python
model=load_model('simplernn_imdb.h5')
```

```python
### preprocess user input
def preprocess_input(text):
    words= text.lower().split()
    encoded_review=[word_index.get(word,2)+3 for word in words]
    encoded_review = [idx if idx<10000 else 2 for idx in encoded_review]
    padded_review=sequence.pad_sequences([encoded_review], maxlen=500)
    return padded_review
```

```python
def decode_review(encoded_review):
    return ''.join([reversed_word_index.get(i-3,'?') for i in encoded_review])
```

```python
### predict sentiment function
def predict_sentiment(review):
    preprocessed=preprocess_input(review)
    prediction=model.predict(preprocessed)
    sentiment= "positive" if prediction[0][0]>0.5 else "negative"
    return sentiment, prediction[0][0]
```

```python
### example 
ex_review="this movie is terrible"
sentiment, likability= predict_sentiment(ex_review)
print(f"review: {ex_review}")
print(f"sentiment:{sentiment}")
print(f"likability :{likability}")
```
