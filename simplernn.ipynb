```python
import numpy as np
import tensorflow as tf
from keras.datasets import imdb
from keras.preprocessing import sequence
from keras.models import Sequential
from keras.layers import Embedding, SimpleRNN, Dense
```

```python


(x_train,y_train),(x_test,y_test)=imdb.load_data(num_words=10000)
```

```python
### padding to get equal len sentence
max_len=500
x_train=sequence.pad_sequences(x_train,maxlen=max_len)
x_test=sequence.pad_sequences(x_test,maxlen=max_len)
```

```python
x_train[0]
```

```python
y_train[0]
```

```python
###creating simplernn model
model=Sequential()
model.add(Embedding(10000,128,input_length=max_len))
model.add(SimpleRNN(128,activation='tanh',dropout=0.2, recurrent_dropout=0.2))
model.add(Dense(1,activation='sigmoid'))
```

```python
model.compile(optimizer=tf.keras.optimizers.Adam(clipnorm=1.0),loss='binary_crossentropy',metrics=['accuracy'])
```

```python
## adding an early check 
from keras.callbacks import EarlyStopping
earlystop=EarlyStopping(monitor='val_loss',patience=5, restore_best_weights=True)
```

```python
###training the model on train dataset
history= model.fit(x_train,y_train,epochs=10,validation_split=0.2,batch_size=32,callbacks=[earlystop])
```
